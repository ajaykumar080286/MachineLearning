```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.metrics import mean_squared_error, r2_score

from sklearn.datasets import make_regression
```

```python
X,y=make_regression(n_samples=100, n_targets=1, n_features=1, n_informative=1, noise=20, random_state=13)
```

```python
plt.scatter(X,y)
```

```python
lr=LinearRegression()
lr.fit(X,y)

print(lr.coef_)
print(lr.intercept_)
y_pred1=lr.predict(X)
```

```python
lr1=Ridge(alpha=10)
lr1.fit(X,y)

print(lr1.coef_)
print(lr1.intercept_)
y_pred2=lr1.predict(X)
```

```python
lr2=Ridge(alpha=100)
lr2.fit(X,y)

print(lr2.coef_)
print(lr2.intercept_)
y_pred3=lr2.predict(X)
```

```python
plt.plot(X,y,"g.")
plt.plot(X,y_pred1,color='red',label='alpha=0')
plt.plot(X,y_pred2,color='blue',label='alpha=10')
plt.plot(X,y_pred3,color='orange',label='alpha=100')
plt.legend()
```

```python
class MeraRidge:

  def __init__(self,alpha):
    self.alpha=alpha
    self.m=None
    self.b=None

  def fit(self,X,y):

    num=0
    dum=0

    for i in range(X.shape[0]):
      num=num+((y[i]-y.mean())*(X[i]-X.mean()))
      dum=dum+(X[i]-X.mean())*(X[i]-X.mean())

    self.m=num/(dum+self.alpha)
    self.b=y.mean()-self.m*X.mean()

    print(self.m,self.b)
    print(num)

  def predict(self,X):
    return self.m*X+self.b
```

```python
lr3=MeraRidge(100)
```

```python
lr3.fit(X,y)
```
