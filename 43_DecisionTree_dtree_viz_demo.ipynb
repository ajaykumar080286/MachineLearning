```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor, plot_tree,DecisionTreeClassifier
from sklearn.datasets import load_iris
```

```python
iris = load_iris()
```

```python
X=iris.data
y=iris.target
```

```python
dt=DecisionTreeRegressor(max_depth=8)
```

```python
dt.fit(X,y)
```

```python
plot_tree(dt)
```

```python
#pip install dtreeviz==1.4.0
```

```python
from dtreeviz.trees import dtreeviz # Import the dtreeviz function from the trees module
import matplotlib.pyplot as plt
```

```python
viz = dtreeviz(dt,
               X,
               y,
               feature_names=iris.feature_names,
               class_names=["setosa", "versicolor", "virginica"])
viz
```

**3. Horizontal Decision Tree**

```python
viz = dtreeviz(dt,
               X,
               y,
               feature_names=iris.feature_names,
               class_names=["setosa", "versicolor", "virginica"],
               scale = 1.5,
               orientation='LR')
viz
```

**4. Show prediction path**

```python
clas = DecisionTreeClassifier()
iris = load_iris()

X_train = iris.data
y_train = iris.target
clas.fit(X_train, y_train)
```

```python
X_input = iris.data[np.random.randint(0, len(iris.data)),:]
```

```python
viz = dtreeviz(clas,
               X_train,
               y_train,
               feature_names=iris.feature_names,
               class_names=["setosa", "versicolor", "virginica"],
               X=X_input)
viz
```

**7. Show just prediction path, nothing else**

```python
clas = DecisionTreeClassifier()
iris = load_iris()

X_train = iris.data
y_train = iris.target
clas.fit(X_train, y_train)

X = iris.data[np.random.randint(0, len(iris.data)),:]

viz = dtreeviz(clas,
               X_train,
               y_train,
               feature_names=iris.feature_names,
               class_names=["setosa", "versicolor", "virginica"],
               X=X,
               show_just_path=True)
viz
```

**8. Prediction Path in Plain english**

```python
from dtreeviz.trees import explain_prediction_path
print(explain_prediction_path(clas, X, feature_names=iris.feature_names, explanation_type="plain_english"))
```

**9. Feature Importance**

```python
print(explain_prediction_path(clas, X, feature_names=iris.feature_names, explanation_type="sklearn_default"))
```

```python
regr = DecisionTreeRegressor(max_depth=1)
```

```python
data_url = "http://lib.stat.cmu.edu/datasets/boston"
raw_df = pd.read_csv(data_url, sep=r"\s+", skiprows=22, header=None)
data = np.hstack([raw_df.values[::2, :], raw_df.values[1::2, :2]])
target = raw_df.values[1::2, 2]

df = pd.DataFrame(data,columns=["CRIM","ZN","INDUS","CHAS","NOX","RM","AGE","DIS","RAD","TAX","PTRATIO","B","LSTAT"])
df['PRICE'] = target
X1=df.iloc[:,0:13]
y1=df.iloc[:,-1]
```

```python
regr.fit(X1, y1)

viz = dtreeviz(regr,
               X1,
               y1,
               feature_names=df.columns,
               scale=2
              )
viz
```

**10. Univariate Regression**

```python
df_cars = pd.read_csv("https://raw.githubusercontent.com/ajaykumar080286/MachineLearning/main/cars.csv", usecols=["mpg","cyl","wt","disp"])
```

```python
df_cars.head(2)
```

```python
X2 = df_cars[["wt"]]
y2=df_cars["mpg"]
```

```python
dt1 = DecisionTreeRegressor(max_depth=3, criterion="absolute_error")
dt1.fit(X2, y2)
```

```python
from dtreeviz.trees import *
fig = plt.figure()
ax = fig.gca()
rtreeviz_univar(dt1, X2, y2, 'wt', 'mpg', ax=ax)
plt.show()
```

```python
from mpl_toolkits.mplot3d import Axes3D
```

```python
X = df_cars[['wt','disp']]
y = df_cars['mpg']

dt = DecisionTreeRegressor(max_depth=3, criterion="absolute_error")
dt.fit(X, y)

figsize = (6,5)
fig = plt.figure(figsize=figsize)
ax = fig.add_subplot(111, projection='3d')

t = rtreeviz_bivar_3D(dt,
                      X, y,
                      feature_names=['Vehicle Weight', 'Horse Power'],
                      target_name='mpg',
                      fontsize=14,
                      elev=20,
                      azim=25,
                      dist=8.2,
                      show={'splits','title'},
                      ax=ax)
plt.show()
```
