```python
import pandas as pd
import numpy as np

from sklearn.cluster import KMeans

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("https://raw.githubusercontent.com/ajaykumar080286/MachineLearning/main/student_clustering.csv")
```

```python
df.head()
```

```python
plt.scatter(df["cgpa"], df["iq"])
```

```python
sns.scatterplot(x=df["cgpa"], y=df["iq"])
```

```python
wcss=[]
for i in range(1,11):
  km=KMeans(n_clusters=i)
  km.fit_predict(df)
  wcss.append(km.inertia_)
```

```python
wcss
```

```python
plt.plot(range(1,11),wcss)
```

```python
km=KMeans(n_clusters=4)
```

```python
X=df.iloc[:,:].values
```

```python
y_pred=km.fit_predict(X)
```

```python
y_pred
```

```python
X[y_pred==0] #rows cluster no 0
```

```python
X[y_pred==3,1] #rows cluster no 0
```

```python
plt.scatter(X[y_pred == 0,0],X[y_pred == 0,1],color='blue')
plt.scatter(X[y_pred == 1,0],X[y_pred == 1,1],color='red')
plt.scatter(X[y_pred == 2,0],X[y_pred == 2,1],color='green')
plt.scatter(X[y_pred == 3,0],X[y_pred == 3,1],color='yellow')
```

**K-Means on 3-D Data**

```python
from sklearn.datasets import make_blobs

centroids = [(-5,-5,5),(5,5,-5),(3.5,-2.5,4),(-2.5,2.5,-4)]
cluster_std = [1,1,1,1]

X,y = make_blobs(n_samples=200,cluster_std=cluster_std,centers=centroids,n_features=3,random_state=1)
```

```python
import plotly.express as px
```

```python
px.scatter_3d(x=X[:,0],y=X[:,1],z=X[:,2])
```

```python
wcss=[]

for i in range(1,10):
  km=KMeans(n_clusters=i)
  km.fit_predict(X)
  wcss.append(km.inertia_)
```

```python
wcss
```

```python
plt.plot(range(1,10),wcss)
```

```python
km=KMeans(n_clusters=4)
```

```python
y_means=km.fit_predict(X)
```

```python
df=pd.DataFrame()
```

```python
df["col0"]=X[:,0]
df["col1"]=X[:,1]
df["col2"]=X[:,2]
df["label"]=y_means
```

```python
fig = px.scatter_3d(df,x='col0', y='col1', z='col2',color='label')
fig.show()
```
