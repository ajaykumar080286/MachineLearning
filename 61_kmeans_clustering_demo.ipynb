```python
import pandas as pd
import numpy as np

from sklearn.cluster import KMeans

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("https://raw.githubusercontent.com/ajaykumar080286/MachineLearning/main/student_clustering.csv")
```

```python
df.head()
```

```python
plt.scatter(df["cgpa"], df["iq"])
```

```python
sns.scatterplot(x=df["cgpa"], y=df["iq"])
```

```python
wcss=[]
for i in range(1,11):
  km=KMeans(n_clusters=i)
  km.fit_predict(df)
  wcss.append(km.inertia_)
```

```python
wcss
```

```python
plt.plot(range(1,11),wcss)
```

```python
km=KMeans(n_clusters=4)
```

```python
X=df.iloc[:,:].values
```

```python
y_pred=km.fit_predict(X)
```

```python
y_pred
```

```python
X[y_pred==0] #rows cluster no 0
```

```python
X[y_pred==3,1] #rows cluster no 0
```

```python
plt.scatter(X[y_pred == 0,0],X[y_pred == 0,1],color='blue')
plt.scatter(X[y_pred == 1,0],X[y_pred == 1,1],color='red')
plt.scatter(X[y_pred == 2,0],X[y_pred == 2,1],color='green')
plt.scatter(X[y_pred == 3,0],X[y_pred == 3,1],color='yellow')
```
