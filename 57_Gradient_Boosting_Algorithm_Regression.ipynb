```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.tree import DecisionTreeRegressor,plot_tree
```

```python
# Create the DataFrame from a dictionary
data = {
    "R&D Spend": [165.0, 101.0, 29.0],
    "Administration": [137.0, 92.0, 127.0],
    "Marketing Spend": [472.0, 250.0, 201.0],
    "Profit": [192.0, 144.0, 91.0]
}

# Specify the index to match your output
df = pd.DataFrame(data, index=[0, 11, 36])

```

```python
df.head()
```

```python
df["df0(x)"]=df["Profit"].mean()
```

```python
df
```

```python
df["ri1"]=df["Profit"]-df["df0(x)"]
```

```python
df
```

```python
dt1=DecisionTreeRegressor(max_depth=1)
```

```python
dt1.fit(df.iloc[:,0:3].values, df["ri1"].values)
```

```python
plot_tree(dt1)
plt.show()
```
