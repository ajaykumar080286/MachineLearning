```python
from sklearn.datasets import make_circles
```

```python
X,_=make_circles(n_samples=500, factor=0.05, noise=0.03, random_state=4)
```

```python
db=DBSCAN(eps=0.1, min_samples=5)
```

```python
clusters=db.fit_predict(X)
```

```python
plt.scatter(X[:,0], X[:,1], c=clusters, cmap='viridis',marker='o')
plt.title('DBSCAN')
plt.xlabel("Featute 0")
plt.ylabel("Featureb 1")
plt.show()
```
