```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from scipy.stats import probplot
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.datasets import make_regression
```

```python
X,y = make_regression(n_samples=4, n_features=1, n_informative=1, n_targets=1,noise=80,random_state=13)
```

```python
plt.scatter(X,y)
```

```python
lr=LinearRegression()
```

```python
lr.fit(X,y)
```

```python
m=lr.coef_
```

```python
b=lr.intercept_
```

```python
plt.scatter(X,y)
plt.plot(X,lr.predict(X),color='red')
```

```python
b=100
lrate=0.1
y_pred=(m*X+b).reshape(4)
y_pred
```

```python
plt.scatter(X,y)
plt.plot(X,lr.predict(X),color='green',label='OLS')
plt.plot(X,y_pred,color='red', label="b={}".format(b))

plt.legend()
plt.show()
```

```python
Slope=-2*np.sum(y-m*X.ravel()-b)
Slope
```

```python
step_size =Slope*lrate
step_size
```

```python
b=b-step_size
b
```

```python
Slope=-2*np.sum(y-m*X.ravel()-b)
step_size=Slope*lrate
b=b-step_size
y_pred2=(m*X+b).reshape(4)

```

```python
plt.scatter(X,y)
plt.plot(X,lr.predict(X),color='green',label='OLS')
plt.plot(X,y_pred2,color='yellow', label="b={}".format(b))
plt.plot(X,y_pred1,color='blue', label="b={}".format(b))
plt.plot(X,y_pred,color='red', label="b=100")
plt.show()
```

```python
b = -100
m = 78.35
lr = 0.01
epochs = 100
for i in range(epochs):

  Slope=-2*np.sum(y-m*X.ravel()-b)
  step_size=Slope*lr
  b=b-step_size
  y_pred=(m*X+b).reshape(4)
  plt.plot(X,y_pred)

plt.scatter(X,y)
```

```python
b = -100
m = 78.35
lr = 0.01

epochs = 100

for i in range(epochs):
  loss_slope = -2 * np.sum(y - m*X.ravel() - b)
  b = b - (lr * loss_slope)

  y_pred = m * X + b
  plt.plot(X,y_pred)

plt.scatter(X,y)
```
