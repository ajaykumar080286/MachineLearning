```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import make_classification
import random
from sklearn.tree import plot_tree
```

```python
X,y = make_classification(n_features=5, n_redundant=0, n_informative=5,n_clusters_per_class=1)
```

```python
X.shape
```

```python
df=pd.DataFrame(X,columns=["col1","col2","col3","col4","col5"])
df["target"]=y
df.head()
```

```python
def row_sample(df,percent):
  return df.sample(int(df.shape[0]*0.1),replace=True)
```

```python
df1=row_sample(df,0.1)
df2=row_sample(df,0.1)
df3=row_sample(df,0.1)
```

```python
dt1=DecisionTreeClassifier()
dt2=DecisionTreeClassifier()
dt3=DecisionTreeClassifier()
```

```python
dt1.fit(df1.iloc[:,0:5],df1.iloc[:,-1])
dt2.fit(df2.iloc[:,0:5],df2.iloc[:,-1])
dt3.fit(df3.iloc[:,0:5],df3.iloc[:,-1])
```

```python
print("pred1",dt1.predict(np.array([-1.043974,0.550450,-1.067709,-1.543306,0.751052]).reshape(1,5)))
print("pred2",dt2.predict(np.array([-1.043974,0.550450,-1.067709,-1.543306,0.751052]).reshape(1,5)))
print("pred3",dt3.predict(np.array([-1.043974,0.550450,-1.067709,-1.543306,0.751052]).reshape(1,5)))
```

```python
plot_tree(dt1)
```

```python
plot_tree(dt2)
```

```python
plot_tree(dt3)
```

**# function for feature sampling**

```python
def col_sample(df,percent):
  col=random.sample(df.columns.tolist()[:-1],int(df.shape[1]*percent))
  new_df=df[col]
  new_df["target"]=df["target"]
  return new_df
```

```python
df_1=col_sample(df,0.5)
df_2=col_sample(df,0.5)
df_3=col_sample(df,0.5)
df_1.head()
```

```python
dt_1=DecisionTreeClassifier()
dt_2=DecisionTreeClassifier()
dt_3=DecisionTreeClassifier()
dt_1.fit(df_1.iloc[:,0:3],df_1.iloc[:,-1])
dt_2.fit(df_2.iloc[:,0:3],df_2.iloc[:,-1])
dt_3.fit(df_3.iloc[:,0:3],df_3.iloc[:,-1])
```

```python
print("pred1",dt_1.predict(np.array([1.148298,0.364606,-1.863308]).reshape(1,3)))
print("pred2",dt_2.predict(np.array([1.148298,0.364606,-1.863308]).reshape(1,3)))
print("pred3",dt_3.predict(np.array([1.148298,0.364606,-1.863308]).reshape(1,3)))
```

***# function for combined sampling***

```python
def combined_sampling(df,row_percent,col_percent):
  new_df = row_sample(df,row_percent)
  return col_sample(new_df,col_percent)
```

```python
df_1=combined_sampling(df,0.1,0.5)
df_2=combined_sampling(df,0.1,0.5)
df_3=combined_sampling(df,0.1,0.5)
df_1.head()
```

```python
dt_1=DecisionTreeClassifier()
dt_2=DecisionTreeClassifier()
dt_3=DecisionTreeClassifier()
dt_1.fit(df_1.iloc[:,0:3],df_1.iloc[:,-1])
dt_2.fit(df_2.iloc[:,0:3],df_2.iloc[:,-1])
dt_3.fit(df_3.iloc[:,0:3],df_3.iloc[:,-1])
```

```python
print("pred1",dt_1.predict(np.array([1.148298,0.364606,-1.863308]).reshape(1,3)))
print("pred2",dt_2.predict(np.array([1.148298,0.364606,-1.863308]).reshape(1,3)))
print("pred3",dt_3.predict(np.array([1.148298,0.364606,-1.863308]).reshape(1,3)))
```
