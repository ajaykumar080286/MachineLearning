```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor, plot_tree
```

```python
np.random.seed(42)
X = np.random.rand(100, 1) - 0.5
y = 3*X[:, 0]**2 + 0.05 * np.random.randn(100)
```

```python
df=pd.DataFrame()
```

```python
df['X']=X.reshape(100)
```

```python
df['y']=y
```

```python
X.shape
```

```python
df
```

```python
plt.scatter(df['X'],df['y'])
plt.title('X vs Y')
```

```python
df.head(4)
```

```python
df["y_pred1"]=df["y"].mean()
```

```python
df.head(5)
```

```python
df["res1"]=df["y"]-df["y_pred1"]
```

```python
df.head(8)
```

```python
plt.scatter(df['X'],df['y'])
plt.plot(df["X"],df["y_pred1"], color="red")
```

```python
dt=DecisionTreeRegressor(max_depth=8)
X1=df["X"].values.reshape(100,1)
y1=df["res1"].values
```

```python
dt.fit(X1,y1)
```

```python
plot_tree(dt)
plt.show()
```

```python
X_test = np.linspace(-0.5, 0.5, 500)
```

```python
y_pred2=0.265458+dt.predict(X_test.reshape(500,1))
```

```python
plt.figure(figsize=(14,4))
plt.subplot(121)
plt.plot(X_test, y_pred2, linewidth=2,color='red')
plt.scatter(df['X'],df['y'])
```

```python
df['pred2'] = 0.265458 + dt.predict(df['X'].values.reshape(100,1))
```

```python
df["res2"]=df["y"]-df["pred2"]
```
