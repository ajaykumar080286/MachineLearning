```python
df=pd.read_csv("https://raw.githubusercontent.com/ajaykumar080286/MachineLearning/main/breast_cancer.csv")
```

```python
df["diagnosis"].value_counts()
```

```python
df.isnull().sum()
```

```python
df.drop(columns=["Unnamed: 32","id"], inplace=True)
```

```python
df.sample()
```

```python
X=df.iloc[:,1:]
y=df.iloc[:,0]
```

```python
lable=LabelEncoder()
y=lable.fit_transform(y)
```

```python
lable.classes_
```

```python
df["diagnosis"].value_counts()
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
scale=StandardScaler()
```

```python
X_train_tf=scale.fit_transform(X_train)
X_test_tf=scale.transform(X_test)
```

```python
X_train.shape
```

```python
knn =KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train_tf,y_train)
```

```python
y_pred=knn.predict(X_test_tf)
```

```python
accuracy_score(y_test,y_pred)
```

```python
knn.classes_
```

```python
knn.predict_proba(X_test_tf).argmax(axis=1)
```
