```python
import pandas as pd
import numpy as np
import scipy.cluster.hierarchy as sch
from sklearn.cluster import  AgglomerativeClustering
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("https://raw.githubusercontent.com/ajaykumar080286/MachineLearning/main/hierarchical-clustering-with-python-and-scikit-learn-shopping-data%20(2).csv")
```

```python
data=df.iloc[:,3:5].values
```

```python
#plt.figure(figsize=(10,7))
#plt.title("Customer DendoGram")

#dend=sch.dendrogram(sch.linkage(data,method="ward"))
```

```python
cluster=AgglomerativeClustering(n_clusters=5, metric="euclidean", linkage="ward")
```

```python
lable=cluster.fit_predict(data)
lable
```

```python
plt.figure(figsize=(10,7))
```

```python
plt.scatter(data[:,0], data[:,1], c=cluster.labels_,cmap="rainbow")
```
